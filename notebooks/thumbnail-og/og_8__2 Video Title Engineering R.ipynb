```python
### DATA PREPROCESSING ###

def rename_features(df):
    'Rename the features.'
    transform_df = df.copy()
    name_map = {
        'snippet.channelId': 'channel.id',
        'snippet.channelTitle': 'channel.name',
        'snippet.categoryId': 'video.category_id',
        'snippet.title': 'video.title',
        'snippet.description': 'video.description',
        'snippet.tags': 'video.tags',
        'snippet.publishedAt': 'video.published_at',
        'snippet.defaultAudioLanguage': 'language.default_audio',
        'snippet.defaultLanguage': 'language.default',
        'contentDetails.caption': 'details.has_captions',
        'contentDetails.duration': 'video.duration_seconds',
        'contentDetails.licensedContent': 'details.licensed_content',
        'contentDetails.contentRating.ytRating': 'details.is_age_restricted',
        'contentDetails.projection': 'details.is_rectangular',
        'contentDetails.definition': 'details.is_not_sd',
        'contentDetails.dimension': 'details.is_2d',
        'statistics.commentCount': 'stats.comment_count',
        'statistics.dislikeCount': 'stats.dislike_count',
        'statistics.favoriteCount': 'stats.favorite_count',
        'statistics.likeCount': 'stats.like_count',
        'statistics.viewCount': 'target.view_count',
        'topicDetails.topicCategories': 'video.topic_categories',
        'liveStreamingDetails.scheduledStartTime': 'live.scheduled_start_time',
        'liveStreamingDetails.scheduledEndTime': 'live.scheduled_end_time',
        'liveStreamingDetails.actualStartTime': 'live.actual_start_time',
        'liveStreamingDetails.actualEndTime': 'live.actual_end_time',
    }
    
    transform_df.rename(columns=name_map, inplace=True)
    return transform_df

def drop_inappropriate_data(df):
    'Drop data inappropriate for this use case.'
    transform_df = df.copy()
    # drop non ascii
    transform_df = transform_df[transform_df['video.title'].apply(lambda x: x.isascii())]
    # drop kinds other than video
    transform_df = transform_df[transform_df['kind'] == 'youtube#video']
    # drop live and upcoming broadcasts
    transform_df = transform_df[transform_df['snippet.liveBroadcastContent'] == 'none']
    return transform_df

def bool_conversion(df):
    'Convert the columns to boolean.'
    transform_df = df.copy()
    # convert caption
    transform_df['details.has_captions'] = transform_df['details.has_captions'].apply(lambda x: True if x == 'true' else False)
    # convert ytRating
    transform_df['details.is_age_restricted'] = transform_df['details.is_age_restricted'].apply(lambda x: False if x is None else True)
    # convert projection
    transform_df['details.is_rectangular'] = transform_df['details.is_rectangular'].apply(lambda x: True if x == 'rectangular' else False)
    # convert dimension
    transform_df['details.is_2d'] = transform_df['details.is_2d'].apply(lambda x: True if x == '2d' else False)
    # convert definition
    transform_df['details.is_not_sd'] = transform_df['details.is_not_sd'].apply(lambda x: True if x != 'sd' else False)
    # convert licensed content
    transform_df['details.licensed_content'] = transform_df['details.licensed_content'].astype(bool)
    return transform_df

def manage_time(df):
    'Manage the time columns.'
    transform_df = df.copy()
    # Convert the 'publishedAt' column to datetime
    transform_df['video.published_at'] = pd.to_datetime(transform_df['video.published_at'])
    # Convert the 'duration' column to seconds
    transform_df['video.duration_seconds'] = transform_df['video.duration_seconds'].apply(lambda x: pd.to_timedelta(x).seconds)
    # Get times to make calculations
    scheduled_start_time = pd.to_datetime(transform_df['live.scheduled_start_time'])
    scheduled_end_time = pd.to_datetime(transform_df['live.scheduled_end_time'])
    actual_start_time = pd.to_datetime(transform_df['live.actual_start_time'])
    actual_end_time = pd.to_datetime(transform_df['live.actual_end_time'])
    # if row has scheduled start time and end time, calculate the scheduled duration
    transform_df['live.scheduled_duration_seconds'] = (scheduled_end_time - scheduled_start_time).dt.total_seconds()
    # if row has scheduled start time and actual start time, calculate the tardiness
    transform_df['live.tardiness_seconds'] = (actual_start_time - scheduled_start_time).dt.total_seconds()
    # if row has actual end time and scheduled end time, calculate the overrun
    transform_df['live.overrun_seconds'] = (actual_end_time - scheduled_end_time).dt.total_seconds()
    # if row has actual start time, 'live' is True, else False
    transform_df['live'] = transform_df['live.scheduled_start_time'].apply(lambda x: True if pd.notnull(x) else False)
    # if row has scheduled start time and actual start time'live_scheduled' is True, else if has actual start time 'live scheduled is False, else None
    transform_df['live.scheduled'] = transform_df['live'] & transform_df['live.actual_start_time'].apply(lambda x: True if pd.notnull(x) else False)
    return transform_df

def drop_unnecessary_cols(df):
    'Drop unnecessary columns.'
    transform_df = df.copy()
    thumbnail_cols = [col for col in transform_df.columns if 'thumbnail' in col]
    region_cols = [col for col in transform_df.columns if 'region' in col]
    localized_cols = [col for col in transform_df.columns if 'local' in col]
    other_cols = ['kind', 'etag', 'snippet.liveBroadcastContent', 'liveStreamingDetails.activeLiveChatId', 'liveStreamingDetails.concurrentViewers']
    drop_cols = thumbnail_cols + region_cols + localized_cols + other_cols
    transform_df.drop(columns=drop_cols, inplace=True)
    # sort the columns alphabetically
    return transform_df

from sklearn.preprocessing import StandardScaler, MinMaxScaler

def target_scaling(df):
    'Scale the target variable.'
    transform_df = df.copy()
    # Group by 'snippet.channelId' and apply scaling
    for _, group in transform_df.groupby('channel.id'):
        transform_df.loc[group.index, 'target.min_max_scaled_view_count'] = MinMaxScaler().fit_transform(group[['target.view_count']])
        transform_df.loc[group.index, 'target.standard_scaled_view_count'] = StandardScaler().fit_transform(group[['target.view_count']])
    transform_df = transform_df.reindex(sorted(transform_df.columns), axis=1)
    return transform_df

def apply_conditions(df):
    'Apply conditions to the dataframe.'
    transform_df = df.copy()
    # drop rows where details.has_captions is False
    transform_df = transform_df[transform_df['details.has_captions']]
    # drop rows where details.is_2d is False
    transform_df = transform_df[transform_df['details.is_2d']]
    # drop rows where details.is_not_sd is False
    transform_df = transform_df[transform_df['details.is_not_sd']]
    # drop rows where details.is_rectangular is False
    transform_df = transform_df[transform_df['details.is_rectangular']]
    # drop rows where none of None, 'en', 'EN' in language.default_audio
    transform_df = transform_df[transform_df['language.default_audio'].apply(lambda x: x in [None, 'en', 'EN'])]
    # drop rows where none of None, 'en', 'EN' in language.default
    transform_df = transform_df[transform_df['language.default'].apply(lambda x: x in [None, 'en', 'EN'])]
    # drop rows where live is True
    transform_df = transform_df[~transform_df['live']]
    return transform_df    
```

```python
### FEATURE ENGINEERING ###

def generate_capitalization_features(data):
    capital_df = data.copy()
    capital_df['num_capital_letters'] = capital_df['video.title'].apply(lambda x: sum(1 for c in x if c.isupper()))
    capital_df['num_all_caps_words'] = capital_df['video.title'].apply(lambda x: sum(1 for c in x.split() if c.isupper()))
    capital_df['num_capitalized_words'] = capital_df['video.title'].apply(lambda x: sum(1 for c in x.split() if c.istitle()))
    capital_df['pct_capital_letters'] = capital_df['num_capital_letters'] / capital_df['video.title'].apply(len)
    capital_df['pct_all_caps_words'] = capital_df['num_all_caps_words'] / capital_df['video.title'].apply(lambda x: len(x.split()))
    capital_df['pct_capitalized_words'] = capital_df['num_capitalized_words'] / capital_df['video.title'].apply(lambda x: len(x.split()))
    capital_df = capital_df.drop(columns=['video.title'])
    return capital_df

def generate_readability_features(data):
    reading_df = data.copy()
    try:
        reading_df['flesch_reading_ease'] = reading_df['video.title'].apply(lambda x: 206.835 - 1.015 * len(x.split()) - 84.6 * sum(1 for c in x if c.lower() in 'aeiouy') / len(x.split()))
        reading_df['flesch_kincaid_grade'] = reading_df['video.title'].apply(lambda x: 0.39 * len(x.split()) + 11.8 * sum(1 for c in x if c.lower() in 'aeiouy') / len(x.split()) - 15.59)
        reading_df['gunning_fog_index'] = reading_df['video.title'].apply(lambda x: 0.4 * (len(x.split()) + 100 * sum(1 for c in x if c in ['.', '!', '?'])) / len(x.split()))
        reading_df['smog_index'] = reading_df['video.title'].apply(lambda x: 1.043 * (30 * sum(1 for c in x if c in ['.', '!', '?']) / len(x.split())) ** 0.5 + 3.1291)
        reading_df['coleman_liau_index'] = reading_df['video.title'].apply(lambda x: 5.88 * sum(1 for c in x if c.isalnum()) / len(x) - 0.296 * 100 / len(x.split()) - 15.8)
        reading_df['automated_readability_index'] = reading_df['video.title'].apply(lambda x: 4.71 * len(x) / len(x.split()) + 0.5 * len(x.split()) - 21.43)
        reading_df['linsear_write_formula'] = reading_df['video.title'].apply(lambda x: (sum(1 for c in x if c.lower() in 'aeiouy') + sum(1 for c in x if c in ['.', '!', '?'])) / len(x.split()) / 2)
        reading_df['dale_chall_readability_score'] = reading_df['video.title'].apply(lambda x: 0.1579 * (100 * sum(1 for c in x if c.isalnum()) / len(x)) + 0.0496 * len(x.split()))
    except ZeroDivisionError:
        reading_df['flesch_reading_ease'] = 0
        reading_df['flesch_kincaid_grade'] = 0    
        reading_df['gunning_fog_index'] = 0
        reading_df['smog_index'] = 0
        reading_df['coleman_liau_index'] = 0
        reading_df['automated_readability_index'] = 0
        reading_df['linsear_write_formula'] = 0
        reading_df['dale_chall_readability_score'] = 0
    reading_df = reading_df.drop(columns=['video.title']).dropna()
    return reading_df

from nltk.tag import pos_tag

from nltk.tokenize import word_tokenize
from collections import Counter

def generate_pos_tags_df(data):
    pos_tags = data['video.title'].apply(lambda x: Counter(tag for word, tag in pos_tag(word_tokenize(x))))

    specific_pos_tags_count_df = pd.DataFrame(list(pos_tags), index=data.index)
    specific_pos_tags_count_df = specific_pos_tags_count_df.fillna(0)

    for col in specific_pos_tags_count_df.columns:
        if not col.isalnum():
            specific_pos_tags_count_df.drop(columns=col, inplace=True)

    generic_pos_tags_count_df = specific_pos_tags_count_df.copy()
    generic_pos_tags_count_df['noun'] = generic_pos_tags_count_df[['NN', 'NNS', 'NNP', 'NNPS']].sum(axis=1)
    generic_pos_tags_count_df['verb'] = generic_pos_tags_count_df[['VB', 'VBD', 'VBG', 'VBN', 'VBP', 'VBZ']].sum(axis=1)
    generic_pos_tags_count_df['adjective'] = generic_pos_tags_count_df[['JJ', 'JJR', 'JJS']].sum(axis=1)
    generic_pos_tags_count_df['adverb'] = generic_pos_tags_count_df[['RB', 'RBR', 'RBS']].sum(axis=1)
    generic_pos_tags_count_df['pronoun'] = generic_pos_tags_count_df[['PRP', 'WP']].sum(axis=1)
    generic_pos_tags_count_df['preposition'] = generic_pos_tags_count_df[['IN', 'TO']].sum(axis=1)
    generic_pos_tags_count_df['conjunction'] = generic_pos_tags_count_df[['CC']].sum(axis=1)
    generic_pos_tags_count_df['interjection'] = generic_pos_tags_count_df[['UH']].sum(axis=1)
    generic_pos_tags_count_df['determiner'] = generic_pos_tags_count_df[['DT', 'PDT', 'WDT']].sum(axis=1)
    generic_pos_tags_count_df = generic_pos_tags_count_df[['noun', 'verb', 'adjective', 'adverb', 'pronoun', 'preposition', 'conjunction', 'interjection', 'determiner']]

    specific_pos_tags_pct_df = specific_pos_tags_count_df.div(specific_pos_tags_count_df.sum(axis=1), axis=0)
    generic_pos_tags_pct_df = generic_pos_tags_count_df.div(generic_pos_tags_count_df.sum(axis=1), axis=0)

    specific_pos_tags_count_df.columns = ['specific_count.' + col for col in specific_pos_tags_count_df.columns]
    specific_pos_tags_pct_df.columns = ['specific_pct.' + col for col in specific_pos_tags_pct_df.columns]
    generic_pos_tags_count_df.columns = ['generic_count.' + col for col in generic_pos_tags_count_df.columns]
    generic_pos_tags_pct_df.columns = ['generic_pct.' + col for col in generic_pos_tags_pct_df.columns]

    all_pos_tags_df = pd.concat([specific_pos_tags_count_df, specific_pos_tags_pct_df, generic_pos_tags_count_df, generic_pos_tags_pct_df], axis=1).fillna(0)
    # add target column
    all_pos_tags_df['target.min_max_scaled_view_count'] = data['target.min_max_scaled_view_count']
    return all_pos_tags_df

import nltk
nltk.download('vader_lexicon')

# score sentiment
from nltk.sentiment.vader import SentimentIntensityAnalyzer
def generate_sentiment_df(data):
    sia = SentimentIntensityAnalyzer()

    sentiment_df = data.copy()
    sentiment_df['sentiment'] = sentiment_df['video.title'].apply(lambda x: sia.polarity_scores(x)['compound'])

    # get sentiment of each word
    def get_word_sentiment(title):
        return [sia.polarity_scores(word)['compound'] for word in title.split()]

    word_sentiment = data['video.title'].apply(get_word_sentiment)
    try:
        sentiment_df['avg_word_sentiment'] = word_sentiment.apply(lambda x: sum(x) / len(x))
    except ZeroDivisionError:
        sentiment_df['avg_word_sentiment'] = 0
    try:
        sentiment_df['max_word_sentiment'] = word_sentiment.apply(max)
        sentiment_df['min_word_sentiment'] = word_sentiment.apply(min)
    except ValueError:
        sentiment_df['max_word_sentiment'] = 0
        sentiment_df['min_word_sentiment'] = 0

    sentiment_df = sentiment_df.drop(columns=['video.title'])
    # pad with zeros
    max_len = max(len(sent) for sent in word_sentiment)
    word_sentiment = word_sentiment.apply(lambda x: x + [0] * (max_len - len(x)))
    word_sentiment_df = pd.DataFrame(word_sentiment.tolist(), index=word_sentiment.index)
    word_sentiment_df.columns = ['word.' + str(i) for i in word_sentiment_df.columns]
    sentiment_df = pd.concat([sentiment_df, word_sentiment_df], axis=1)
    return sentiment_df

def get_feature_significance(train_data_X, train_data_y):
    lasso_estimator = fit_get_lasso_estimator(train_data_X, train_data_y)
    lasso_coefficients = pd.Series(lasso_estimator.coef_, index=train_data_X.columns, name='coefficient')

    rf_estimator = fit_get_rf_estimator(train_data_X, train_data_y)
    rf_importances = pd.Series(rf_estimator.feature_importances_, index=train_data_X.columns, name='importance')

    features = pd.concat([lasso_coefficients, rf_importances], axis=1)
    features['score'] = features['coefficient'] * features['importance']
    features = features.sort_values('score', ascending=False)
    return features
```

```python
### MODELING ###

from sklearn.preprocessing import MinMaxScaler

def scale_data(train_data, test_data, scaler=None):
    if not scaler:
        scaler = MinMaxScaler()
    train_data = pd.DataFrame(scaler.fit_transform(train_data),
                              columns=train_data.columns, 
                              index=train_data.index)
    test_data = pd.DataFrame(scaler.transform(test_data),
                             columns=test_data.columns, 
                             index=test_data.index)
    return train_data, test_data

# add to path
import sys
import os
sys.path.append(os.path.abspath(os.path.join('..', 'src')))
from Selector import CombineCorrelatedFeaturesSelector

def combine_correlated_features(train_data, test_data, threshold=1):
    ccfs = CombineCorrelatedFeaturesSelector(threshold=threshold)
    train_data = ccfs.fit_transform(train_data)
    test_data = ccfs.transform(test_data)
    return train_data, test_data
    
from Selector import MutualInforegressionSelector

def select_by_mutual_inforegression(train_data, train_target, test_data, threshold=0):
    mirs = MutualInforegressionSelector(threshold=threshold)
    train_data = mirs.fit_transform(train_data, train_target)
    test_data = mirs.transform(test_data)
    return train_data, test_data

from Selector import VIFSelector

def select_by_variance_inflation_factor(train_data, test_data, threshold=10):
    vifs = VIFSelector(threshold=threshold)
    train_data = vifs.fit_transform(train_data)
    test_data = vifs.transform(test_data)
    return train_data, test_data

from Selector import CooksDistanceInfluenceSelector

def remove_influential_points(train_data, train_target):
    cd = CooksDistanceInfluenceSelector()
    train_data = cd.fit_transform(train_data, train_target)
    train_target = train_target.loc[train_data.index]
    return train_data, train_target

from sklearn.linear_model import Lasso
from sklearn.model_selection import RandomizedSearchCV

def fit_get_lasso_estimator(train_data, train_target):
    print('Lasso')
    lasso_params = { 'alpha' : [.1**i for i in range(10)] } 
    lasso = Lasso(max_iter=10000)
    lasso_search = RandomizedSearchCV(lasso, param_distributions=lasso_params, n_iter=8, cv=3, n_jobs=-1, verbose=2, random_state=42)
    lasso_search.fit(train_data, train_target)
    lasso_best_estimator = lasso_search.best_estimator_
    print(lasso_search.best_params_)
    print(lasso_search.best_score_)
    return lasso_best_estimator

from sklearn.ensemble import RandomForestRegressor

def fit_get_rf_estimator(train_data, train_target):
    print('Random Forest')
    rf_params = {
        'n_estimators': [2**i for i in range(13)],
        'max_features': ['sqrt', 'log2'],
        'max_depth': [2**i for i in range(11)],
        'min_samples_split': [2**i for i in range(1,9)],
        'min_samples_leaf': [2**i for i in range(1,9)]
    }                      
    rf = RandomForestRegressor()
    rf_search = RandomizedSearchCV(rf, param_distributions=rf_params, n_iter=16, cv=4, n_jobs=-1, verbose=2, random_state=42)
    rf_search.fit(train_data, train_target)
    rf_best_estimator = rf_search.best_estimator_
    print(rf_search.best_params_)
    print(rf_search.best_score_)
    return rf_best_estimator

def get_train_test_X_y(train_data, test_data, target_column_name):
    X_train = train_data.drop(columns=[target_column_name])
    y_train = train_data[target_column_name]
    X_test = test_data.drop(columns=[target_column_name])
    y_test = test_data[target_column_name]
    return X_train, y_train, X_test, y_test
```
