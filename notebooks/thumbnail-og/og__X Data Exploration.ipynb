```python
data_cols = ['id', 'snippet_publishedAt', 'snippet_channelId', 'snippet_title', 'snippet_description', 'snippet_categoryId', 'snippet_channelTitle', 'snippet_tags', 
             'statistics_viewCount', 'statistics_likeCount', 'statistics_commentCount', 'contentDetails_duration', 'contentDetails_definition', 'contentDetails_caption',
             'topicDetails_topicCategories', 'timestamp', 'query']
data_df = video_df.copy()
data_df = data_df[data_cols]

category_map = {1: 'Film & Animation', 2: 'Autos & Vehicles', 10: 'Music', 15: 'Pets & Animals', 17: 'Sports', 18: 'Short Movies', 19: 'Travel & Events', 20: 'Gaming',
                21: 'Videoblogging', 22: 'People & Blogs', 23: 'Comedy', 24: 'Entertainment', 25: 'News & Politics', 26: 'Howto & Style', 27: 'Education', 28: 'Science & Technology',
                29: 'Nonprofits & Activism', 30: 'Movies', 31: 'Anime/Animation', 32: 'Action/Adventure', 33: 'Classics', 34: 'Comedy', 35: 'Documentary', 36: 'Drama',
                37: 'Family', 38: 'Foreign', 39: 'Horror', 40: 'Sci-Fi/Fantasy', 41: 'Thriller', 42: 'Shorts', 43: 'Shows', 44: 'Trailers'}

data_df['snippet_categoryId'] = data_df['snippet_categoryId'].astype(int).map(category_map)

data_df
```

```python
import warnings

warnings.simplefilter(action='ignore', category=FutureWarning)
warnings.simplefilter(action='ignore', category=DeprecationWarning)

import numpy as np
import pandas as pd

pd.options.display.float_format = '{:.2f}'.format
pd.set_option('display.width', 256)

# Convert all columns to numeric where possible
group_data_df = data_df.apply(pd.to_numeric, errors='ignore')

# Group by 'Category' and calculate the mean of numeric columns
grouped_counts = group_data_df[['snippet_categoryId', 'statistics_viewCount']].groupby('snippet_categoryId').apply(lambda x: x.select_dtypes(include='number').count())
grouped_means = group_data_df[['snippet_categoryId', 'statistics_viewCount']].groupby('snippet_categoryId').apply(lambda x: x.select_dtypes(include='number').mean())
grouped_medians = group_data_df[['snippet_categoryId', 'statistics_viewCount']].groupby('snippet_categoryId').apply(lambda x: x.select_dtypes(include='number').median())
grouped_skew = group_data_df[['snippet_categoryId', 'statistics_viewCount']].groupby('snippet_categoryId').apply(lambda x: x.select_dtypes(include='number').skew())
grouped_kurt = group_data_df[['snippet_categoryId', 'statistics_viewCount']].groupby('snippet_categoryId').apply(lambda x: x.select_dtypes(include='number').kurt())
grouped_data = pd.concat([grouped_counts, grouped_means, grouped_medians, grouped_skew, grouped_kurt], axis=1)
grouped_data.columns = ['Count', 'Mean', 'Median', 'Skew', 'Kurtosis']
grouped_data.sort_values('Mean', ascending=False)
```

```python
# Convert all columns to numeric where possible
describe_data_df = data_df.apply(pd.to_numeric, errors='ignore')
# get dataframe of descriptive statistics for view_counts of each category
describe_data_gb = describe_data_df.groupby('snippet_categoryId')['statistics_viewCount'].describe()
# Add .1 and .9 percentiles
describe_data_gb['10%'] = describe_data_df.groupby('snippet_categoryId')['statistics_viewCount'].quantile(.1)
describe_data_gb['1%'] = describe_data_df.groupby('snippet_categoryId')['statistics_viewCount'].quantile(.01)
describe_data_gb['90%'] = describe_data_df.groupby('snippet_categoryId')['statistics_viewCount'].quantile(.9)
# Add median
describe_data_gb['median'] = describe_data_df.groupby('snippet_categoryId')['statistics_viewCount'].median()

describe_data_gb.sort_values('1%', ascending=False)
```

It seems to me as though Science and Technology is a "guaranteed" viewership, to an extent. Let's look at what science and technology videos look like.

Let's get tfidf vectorizations of the title, both with and without max features.

```python
science_df = data_df[data_df['snippet_categoryId'] == 'Science & Technology']

from sklearn.feature_extraction.text import TfidfVectorizer

# Create a TfidfVectorizer object
maxn_science_title_tfidf = TfidfVectorizer(stop_words='english', max_features=int(np.sqrt(len(science_df))))
full_science_title_tfidf = TfidfVectorizer(stop_words='english')

# Fit and transform the vectorizer on the 'snippet_title' column
maxn_science_title_tfidf_matrix = maxn_science_title_tfidf.fit_transform(science_df['snippet_title'])
full_science_title_tfidf_matrix = full_science_title_tfidf.fit_transform(science_df['snippet_title'])

maxn_science_title_tfidf_df = pd.DataFrame(maxn_science_title_tfidf_matrix.toarray(), columns=maxn_science_title_tfidf.get_feature_names_out(), index=science_df['id'])
full_science_title_tfidf_df = pd.DataFrame(full_science_title_tfidf_matrix.toarray(), columns=full_science_title_tfidf.get_feature_names_out(), index=science_df['id'])

maxn_science_title_tfidf_df.shape[1], full_science_title_tfidf_df.shape[1]
```

Let's POC with Lasso, XGB for feature selection

```python
# suppresse ConvergenceWarning
import warnings
from sklearn.exceptions import ConvergenceWarning
warnings.simplefilter(action='ignore', category=ConvergenceWarning)

# import lasso, xgb
from sklearn.linear_model import Lasso
from xgboost import XGBRegressor

from sklearn.model_selection import train_test_split

maxn_X_train, maxn_X_test, maxn_y_train, maxn_y_test = train_test_split(maxn_science_title_tfidf_df, science_df['statistics_viewCount'], test_size=0.2, random_state=42)

maxn_lasso = Lasso(alpha=0.1)
maxn_xgb = XGBRegressor(objective='reg:squarederror')

maxn_lasso.fit(maxn_X_train, maxn_y_train)
maxn_xgb.fit(maxn_X_train, maxn_y_train)

maxn_lasso_score = maxn_lasso.score(maxn_X_test, maxn_y_test)
maxn_xgb_score = maxn_xgb.score(maxn_X_test, maxn_y_test)

print(f'Max n scores.', 'Lasso', maxn_lasso_score, 'XGB', maxn_xgb_score)

full_X_train, full_X_test, full_y_train, full_y_test = train_test_split(full_science_title_tfidf_df, science_df['statistics_viewCount'], test_size=0.2, random_state=42)

full_lasso = Lasso(alpha=0.1)
full_xgb = XGBRegressor(objective='reg:squarederror')

full_lasso.fit(full_X_train, full_y_train)
full_xgb.fit(full_X_train, full_y_train)

full_lasso_score = full_lasso.score(full_X_test, full_y_test)
full_xgb_score = full_xgb.score(full_X_test, full_y_test)

print(f'Full scores', 'Lasso', full_lasso_score, 'XGB', full_xgb_score)
```

```python
maxn_coef_df = pd.DataFrame(maxn_lasso.coef_, index=maxn_science_title_tfidf.get_feature_names_out(), columns=['Coefficient'])
maxn_imp_df = pd.DataFrame(maxn_xgb.feature_importances_, index=maxn_science_title_tfidf.get_feature_names_out(), columns=['Importance'])
maxn_features_df = pd.concat([maxn_coef_df, maxn_imp_df], axis=1)
print('Max n coef shape', maxn_coef_df.shape, 'Max n imp shape', maxn_imp_df.shape, 'Max n features shape', maxn_features_df.shape)

full_coef_df = pd.DataFrame(full_lasso.coef_, index=full_science_title_tfidf.get_feature_names_out(), columns=['Coefficient'])
full_imp_df = pd.DataFrame(full_xgb.feature_importances_, index=full_science_title_tfidf.get_feature_names_out(), columns=['Importance'])
full_features_df = pd.concat([full_coef_df, full_imp_df], axis=1)
print('Full coef shape', full_coef_df.shape, 'Full imp shape', full_imp_df.shape, 'Full features shape', full_features_df.shape)
```

```python
maxn_coef_keep_df = maxn_coef_df[maxn_coef_df['Coefficient'] != 0]
maxn_coef_tfidf_df = maxn_science_title_tfidf_df[maxn_coef_keep_df.index]
maxn_coef_X_train, maxn_coef_X_test, maxn_coef_y_train, maxn_coef_y_test = train_test_split(maxn_coef_tfidf_df, science_df['statistics_viewCount'], test_size=0.2, random_state=42)
maxn_coef_lasso = Lasso(alpha=0.1)
maxn_coef_xgb = XGBRegressor(objective='reg:squarederror')
maxn_coef_lasso.fit(maxn_coef_X_train, maxn_coef_y_train)
maxn_coef_xgb.fit(maxn_coef_X_train, maxn_coef_y_train)
maxn_coef_lasso_score = maxn_coef_lasso.score(maxn_coef_X_test, maxn_coef_y_test)
maxn_coef_xgb_score = maxn_coef_xgb.score(maxn_coef_X_test, maxn_coef_y_test)
print('Max n coef features:', len(maxn_coef_tfidf_df.columns), 'Lasso', maxn_coef_lasso_score, 'XGB', maxn_coef_xgb_score)
    
maxn_imp_keep_df = maxn_imp_df[maxn_imp_df['Importance'] != 0]
maxn_imp_tfidf_df = maxn_science_title_tfidf_df[maxn_imp_keep_df.index]
maxn_imp_X_train, maxn_imp_X_test, maxn_imp_y_train, maxn_imp_y_test = train_test_split(maxn_imp_tfidf_df, science_df['statistics_viewCount'], test_size=0.2, random_state=42)
maxn_imp_lasso = Lasso(alpha=0.1)
maxn_imp_xgb = XGBRegressor(objective='reg:squarederror')
maxn_imp_lasso.fit(maxn_imp_X_train, maxn_imp_y_train)
maxn_imp_xgb.fit(maxn_imp_X_train, maxn_imp_y_train)
maxn_imp_lasso_score = maxn_imp_lasso.score(maxn_imp_X_test, maxn_imp_y_test)
maxn_imp_xgb_score = maxn_imp_xgb.score(maxn_imp_X_test, maxn_imp_y_test)
print('Max n imp features:', len(maxn_imp_tfidf_df.columns), 'Lasso', maxn_imp_lasso_score, 'XGB', maxn_imp_xgb_score)

maxn_features_keep_df = maxn_features_df[maxn_features_df['Coefficient'] * maxn_features_df['Importance'] != 0]
maxn_features_tfidf_df = maxn_science_title_tfidf_df[maxn_features_keep_df.index]
maxn_features_X_train, maxn_features_X_test, maxn_features_y_train, maxn_features_y_test = train_test_split(maxn_features_tfidf_df, science_df['statistics_viewCount'], test_size=0.2, random_state=42)
maxn_features_lasso = Lasso(alpha=0.1)
maxn_features_xgb = XGBRegressor(objective='reg:squarederror')
maxn_features_lasso.fit(maxn_features_X_train, maxn_features_y_train)
maxn_features_xgb.fit(maxn_features_X_train, maxn_features_y_train)
maxn_features_lasso_score = maxn_features_lasso.score(maxn_features_X_test, maxn_features_y_test)
maxn_features_xgb_score = maxn_features_xgb.score(maxn_features_X_test, maxn_features_y_test)
print('Max n features:', len(maxn_features_tfidf_df.columns), 'Lasso', maxn_features_lasso_score, 'XGB', maxn_features_xgb_score)
print()

full_coef_keep_df = full_coef_df[full_coef_df['Coefficient'] != 0]
full_coef_tfidf_df = full_science_title_tfidf_df[full_coef_keep_df.index]
full_coef_X_train, full_coef_X_test, full_coef_y_train, full_coef_y_test = train_test_split(full_coef_tfidf_df, science_df['statistics_viewCount'], test_size=0.2, random_state=42)
full_coef_lasso = Lasso(alpha=0.1)
full_coef_xgb = XGBRegressor(objective='reg:squarederror')
full_coef_lasso.fit(full_coef_X_train, full_coef_y_train)
full_coef_xgb.fit(full_coef_X_train, full_coef_y_train)
full_coef_lasso_score = full_coef_lasso.score(full_coef_X_test, full_coef_y_test)
full_coef_xgb_score = full_coef_xgb.score(full_coef_X_test, full_coef_y_test)
print('Full coef features:', len(full_coef_tfidf_df.columns), 'Lasso', full_coef_lasso_score, 'XGB', full_coef_xgb_score)

full_imp_keep_df = full_imp_df[full_imp_df['Importance'] != 0]
full_imp_tfidf_df = full_science_title_tfidf_df[full_imp_keep_df.index]
full_imp_X_train, full_imp_X_test, full_imp_y_train, full_imp_y_test = train_test_split(full_imp_tfidf_df, science_df['statistics_viewCount'], test_size=0.2, random_state=42)
full_imp_lasso = Lasso(alpha=0.1)
full_imp_xgb = XGBRegressor(objective='reg:squarederror')
full_imp_lasso.fit(full_imp_X_train, full_imp_y_train)
full_imp_xgb.fit(full_imp_X_train, full_imp_y_train)
full_imp_lasso_score = full_imp_lasso.score(full_imp_X_test, full_imp_y_test)
full_imp_xgb_score = full_imp_xgb.score(full_imp_X_test, full_imp_y_test)
print('Full imp features:', len(full_imp_tfidf_df.columns), 'Lasso', full_imp_lasso_score, 'XGB', full_imp_xgb_score)

full_features_keep_df = full_features_df[full_features_df['Coefficient'] * full_features_df['Importance'] != 0]
full_features_tfidf_df = full_science_title_tfidf_df[full_features_keep_df.index]
full_features_X_train, full_features_X_test, full_features_y_train, full_features_y_test = train_test_split(full_features_tfidf_df, science_df['statistics_viewCount'], test_size=0.2, random_state=42)
full_features_lasso = Lasso(alpha=0.1)
full_features_xgb = XGBRegressor(objective='reg:squarederror')
full_features_lasso.fit(full_features_X_train, full_features_y_train)
full_features_xgb.fit(full_features_X_train, full_features_y_train)
full_features_lasso_score = full_features_lasso.score(full_features_X_test, full_features_y_test)
full_features_xgb_score = full_features_xgb.score(full_features_X_test, full_features_y_test)
print('Full features:', len(full_features_tfidf_df.columns), 'Lasso', full_features_lasso_score, 'XGB', full_features_xgb_score)
```

Reviewing this data, it's very clear that BuzzFeed dominates the rankings, so, in a way, we're looking at what words predict BuzzFeed videos. Let's try creator-adjusted percentiles.

```python
adj_df = data_df.copy()
# adjust view counts to be channelId percentiled
adj_df['statistics_viewCount'] = adj_df['statistics_viewCount'].fillna(0).astype(int)
adj_df['adjViewCount'] = adj_df.groupby('snippet_channelId')['statistics_viewCount'].transform(lambda x: (x - x.min()) / (x.max() - x.min()))
adj_df.sort_values('adjViewCount', ascending=False)
adj_df.dropna(subset=['adjViewCount'], inplace=True)
adj_df
```

Delightful. Let's try to model.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

tfidf = TfidfVectorizer(stop_words='english', min_df=0.01, max_df=1.0)
tfidf_matrix = tfidf.fit_transform(adj_df['snippet_title'])
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf.get_feature_names_out(), index=adj_df['id'])
tfidf_df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Lasso

X_train, X_test, y_train, y_test = train_test_split(tfidf_df, adj_df['adjViewCount'], test_size=0.2, random_state=42)

lasso = Lasso(alpha=0.0001)

lasso.fit(X_train, y_train)

lasso_score = lasso.score(X_test, y_test)

print('Lasso', lasso_score)
```

```python
from xgboost import XGBRegressor

xgb = XGBRegressor(objective='reg:squarederror')

xgb.fit(X_train, y_train)

xgb_score = xgb.score(X_test, y_test)

print('XGB', xgb_score)
```

```python
import matplotlib.pyplot as plt

coef_df = pd.DataFrame(lasso.coef_, index=tfidf.get_feature_names_out(), columns=['Coefficient'])
imp_df = pd.DataFrame(xgb.feature_importances_, index=tfidf.get_feature_names_out(), columns=['Importance'])
feature_df = pd.concat([coef_df, imp_df], axis=1).fillna(0)
feature_df = feature_df[feature_df['Coefficient'] * feature_df['Importance'] != 0].sort_values('Coefficient', ascending=True)

colors = feature_df['Importance']
fig, ax = plt.subplots(figsize=(10, 10))

feature_df['Coefficient'].plot(kind='barh', ax=ax, color=plt.cm.viridis(colors / max(colors)))
plt.show()
```
